# Content-Based Course Recommendation System

This notebook implements a content-based filtering approach for recommending educational courses based on course descriptions and metadata.

## Setting up the environment

First, let's import the necessary libraries for our recommendation system.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# NLP libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Download necessary NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

## Loading and Exploring the Dataset

We'll use a dataset of courses with descriptions, categories, and other metadata.

```python
# Load the course dataset
courses_df = pd.read_csv('data/course_genre.csv')

# Display the first few rows
courses_df.head()
```

```python
# Check the dataset info
courses_df.info()
```

```python
# Check for missing values
courses_df.isnull().sum()
```

## Data Preprocessing

We need to preprocess the text data (course descriptions) before we can use it for content-based recommendations.

```python
# Text preprocessing function
def preprocess_text(text):
    if isinstance(text, str):
        # Convert to lowercase
        text = text.lower()
        
        # Tokenization
        tokens = word_tokenize(text)
        
        # Remove stopwords and non-alphabetic tokens
        stop_words = set(stopwords.words('english'))
        tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
        
        # Lemmatization
        lemmatizer = WordNetLemmatizer()
        tokens = [lemmatizer.lemmatize(word) for word in tokens]
        
        return ' '.join(tokens)
    else:
        return ''
```

```python
# Apply preprocessing to the course descriptions
courses_df['processed_description'] = courses_df['description'].apply(preprocess_text)

# Check the result
courses_df[['description', 'processed_description']].head(2)
```

## Feature Extraction using TF-IDF

We'll convert the preprocessed text into numerical features using TF-IDF vectorization.

```python
# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=5000)

# Generate TF-IDF vectors for course descriptions
tfidf_matrix = tfidf_vectorizer.fit_transform(courses_df['processed_description'])

# Get feature names (words)
feature_names = tfidf_vectorizer.get_feature_names_out()

# Show the shape of the TF-IDF matrix
print(f"TF-IDF matrix shape: {tfidf_matrix.shape}")
```

## Computing Course Similarity

We'll use cosine similarity to measure how similar courses are to each other based on their descriptions.

```python
# Calculate cosine similarity between all courses
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Create a DataFrame for the similarity matrix
cosine_sim_df = pd.DataFrame(cosine_sim, index=courses_df['course_id'], columns=courses_df['course_id'])

# Display a small portion of the similarity matrix
cosine_sim_df.iloc[:5, :5]
```

## Building the Recommendation Function

Now we'll create a function that returns the top N most similar courses to a given course.

```python
def get_course_recommendations(course_id, similarity_matrix, courses_data, top_n=5):
    """
    Function to get top N course recommendations for a given course
    
    Parameters:
    course_id (int): ID of the course for which to get recommendations
    similarity_matrix (pd.DataFrame): DataFrame containing course similarity scores
    courses_data (pd.DataFrame): DataFrame containing course information
    top_n (int): Number of recommendations to return
    
    Returns:
    pd.DataFrame: DataFrame with recommended courses and their similarity scores
    """
    # Get similarity scores for the specified course
    course_similarities = similarity_matrix[course_id]
    
    # Sort courses by similarity score in descending order
    similar_courses = course_similarities.sort_values(ascending=False)
    
    # Exclude the input course itself (which will have similarity=1)
    similar_courses = similar_courses.drop(course_id)
    
    # Get the top N similar courses
    top_similar_courses = similar_courses.head(top_n)
    
    # Create a dataframe of the recommended courses with similarity scores
    recommendations = courses_data[courses_data['course_id'].isin(top_similar_courses.index)].copy()
    recommendations['similarity_score'] = recommendations['course_id'].apply(lambda x: top_similar_courses[x])
    
    # Sort by similarity score in descending order
    recommendations = recommendations.sort_values('similarity_score', ascending=False)
    
    return recommendations[['course_id', 'title', 'category', 'similarity_score']]
```

## Testing the Recommendation System

Let's test our recommendation system with a few example courses.

```python
# Select a test course
test_course_id = courses_df['course_id'].iloc[42]  # Arbitrarily choosing a course

# Get the course details
test_course = courses_df[courses_df['course_id'] == test_course_id]
print(f"Test Course: {test_course['title'].values[0]}")
print(f"Category: {test_course['category'].values[0]}")
print(f"\nDescription: {test_course['description'].values[0]}\n")
```

```python
# Get recommendations for the test course
recommendations = get_course_recommendations(test_course_id, cosine_sim_df, courses_df, top_n=5)

# Display the recommendations
print("Recommended Courses:")
recommendations
```

## Enhancing the Recommendation System

Now let's improve our system by incorporating additional course features beyond just the description.

```python
# Create a combined feature by joining description with category and other metadata
courses_df['combined_features'] = courses_df['processed_description'] + ' ' + \
                                 courses_df['category'].apply(lambda x: preprocess_text(x)) + ' ' + \
                                 courses_df['level'].apply(lambda x: preprocess_text(x))

# Apply TF-IDF on the combined features
enhanced_tfidf_vectorizer = TfidfVectorizer(max_features=5000)
enhanced_tfidf_matrix = enhanced_tfidf_vectorizer.fit_transform(courses_df['combined_features'])

# Calculate enhanced cosine similarity
enhanced_cosine_sim = cosine_similarity(enhanced_tfidf_matrix, enhanced_tfidf_matrix)
enhanced_cosine_sim_df = pd.DataFrame(enhanced_cosine_sim, index=courses_df['course_id'], columns=courses_df['course_id'])
```

```python
# Get enhanced recommendations for the same test course
enhanced_recommendations = get_course_recommendations(test_course_id, enhanced_cosine_sim_df, courses_df, top_n=5)

# Display the enhanced recommendations
print("Enhanced Recommendations (including category and level):")
enhanced_recommendations
```

## Evaluating the Recommendation Quality

Let's evaluate the quality of our recommendations by calculating precision for courses within the same category.

```python
def calculate_category_precision(course_id, recommendations_df, courses_data):
    """
    Calculate the precision based on category matching
    """
    # Get the category of the input course
    input_category = courses_data[courses_data['course_id'] == course_id]['category'].values[0]
    
    # Count how many recommended courses are in the same category
    same_category_count = sum(recommendations_df['category'] == input_category)
    
    # Calculate precision
    precision = same_category_count / len(recommendations_df)
    
    return precision
```

```python
# Calculate precision for our test course using both recommendation methods
basic_precision = calculate_category_precision(test_course_id, recommendations, courses_df)
enhanced_precision = calculate_category_precision(test_course_id, enhanced_recommendations, courses_df)

print(f"Basic Content-Based Recommendation Precision: {basic_precision:.2f}")
print(f"Enhanced Content-Based Recommendation Precision: {enhanced_precision:.2f}")
```

## User-Based Recommendations

Let's extend our system to provide recommendations for a specific user based on their course history.

```python
# Load user enrollment data
user_courses_df = pd.read_csv('data/ratings.csv')
user_courses_df.head()
```

```python
def get_user_recommendations(user_id, user_courses_data, similarity_matrix, courses_data, top_n=5):
    """
    Get course recommendations for a user based on their course history
    """
    # Get the courses the user has taken
    user_course_ids = user_courses_data[user_courses_data['user_id'] == user_id]['course_id'].tolist()
    
    # If the user has no course history, return popular courses
    if not user_course_ids:
        return courses_data.sort_values('popularity', ascending=False).head(top_n)
    
    # Initialize a dictionary to store course recommendations and their scores
    course_scores = {}
    
    # For each course the user has taken
    for course_id in user_course_ids:
        # Get similar courses
        similar_courses = similarity_matrix[course_id].drop(user_course_ids)
        
        # Add similarity scores to the dictionary
        for sim_course_id, score in similar_courses.items():
            if sim_course_id in course_scores:
                # If the course is already in the dictionary, take the maximum similarity score
                course_scores[sim_course_id] = max(course_scores[sim_course_id], score)
            else:
                course_scores[sim_course_id] = score
    
    # Convert the dictionary to a Series and sort by score
    course_scores = pd.Series(course_scores).sort_values(ascending=False)
    
    # Get the top N recommendations
    top_recommendations = course_scores.head(top_n)
    
    # Get details of the recommended courses
    recommendations = courses_data[courses_data['course_id'].isin(top_recommendations.index)].copy()
    recommendations['similarity_score'] = recommendations['course_id'].apply(lambda x: top_recommendations[x])
    
    # Sort by similarity score
    recommendations = recommendations.sort_values('similarity_score', ascending=False)
    
    return recommendations[['course_id', 'title', 'category', 'similarity_score']]
```

```python
# Test user recommendations
test_user_id = user_courses_df['user_id'].iloc[5]  # Arbitrarily choosing a user

# Get the user's course history
user_history = user_courses_df[user_courses_df['user_id'] == test_user_id]
user_course_ids = user_history['course_id'].tolist()

print(f"User {test_user_id} has taken {len(user_course_ids)} courses:")
for course_id in user_course_ids:
    course_title = courses_df[courses_df['course_id'] == course_id]['title'].values[0]
    print(f"- {course_title}")
```

```python
# Get recommendations for the test user
user_recommendations = get_user_recommendations(test_user_id, user_courses_df, enhanced_cosine_sim_df, courses_df, top_n=5)

# Display the recommendations
print(f"\nRecommended courses for user {test_user_id}:")
user_recommendations
```
